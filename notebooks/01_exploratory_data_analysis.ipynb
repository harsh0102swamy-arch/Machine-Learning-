# 01 - Exploratory Data Analysis & Baseline Model

This notebook demonstrates how to load datasets, explore distributions, and train a baseline model using the project's modular `src/` modules.

```python
import sys
from pathlib import Path

# Add project root to sys.path
project_root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
if str(project_root) not in sys.path:
    sys.path.append(str(project_root))

import pandas as pd
import numpy as np
from src.data_loader import generate_synthetic_data, split_data
from src.features import build_preprocessor
from src.model import create_pipeline, evaluate_model
```

## 1. Load / Generate Data

```python
X, y = generate_synthetic_data(n_samples=1000, n_features=8, random_state=42)
df = X.copy()
df['target'] = y
df.head()
```

## 2. Summary Statistics

```python
df.describe().T
```

## 3. Train Baseline Model & Evaluate

```python
X_train, X_test, y_train, y_test = split_data(X, y, test_size=0.2, random_state=42)
preprocessor = build_preprocessor(numeric_features=list(X.columns))
model = create_pipeline(preprocessor=preprocessor, random_state=42)
model.fit(X_train, y_train)

metrics = evaluate_model(model, X_test, y_test)
for metric, val in metrics.items():
    if metric != 'classification_report':
        print(f"{metric:15s}: {val:.4f}")
```
